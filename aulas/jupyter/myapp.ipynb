```python
import ipywidgets as widgets
from ipywidgets import interact
from IPython.display import display

def modelo(alpha):
    return alpha * 2
slider = interact(modelo, alpha=(0, 10))

button = widgets.Button(description="Clique")
def on_click(b):
    print("Clicado!")
button.on_click(on_click)

textbox = widgets.Text(
    placeholder='Digite algo',
    description='Texto:'
)
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

url = "https://github.com/mauriciodev/progcart/raw/refs/heads/main/dados/pessoas.gpkg"
gdf = gpd.read_file(url)
df_output = widgets.Output()
with df_output:
    display(gdf)

ui = widgets.VBox([
    button,textbox, df_output
])

display(ui)
```
